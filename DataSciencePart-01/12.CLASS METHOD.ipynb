```python
class pwskills1:
    def __init__(self,name,email):
        self.name=name
        self.email=email
    @classmethod
    def details(cls,name,email):
        return cls(name,email)
        
    def student_details(self):
        print(self.name,self.email)
```

```python
pw=pwskills1("kavan","kavan45@gmail.com")
```
