```python
def test():
    print("this is the start of my fun")
    print("this is my fun to test")
    print(4+5)
    print("this is the end of my fun")
```

```python
test()
```

```python
def decor(func):
    def inner_decor():
        print("this is my start of my fun")
        func()
        print("this is my end of my fun")
    return inner_decor
```

```python
@decor
def test1():
    print(6+7)
```

```python
test1()
```

```python
import time

def timer_test(func):
    def timer_test_inner():
        start=time.time()
        func()
        end=time.time()
        print(end-start)
    return timer_test_inner
        
```

```python
def test():
    print(45+78)
    
```

```python
test()
```

```python
@timer_test
def test():
    print(45+78)
```

```python
test()
```

```python
@timer_test
def test():
    for i in range(10000000):
        pass
```

```python
test()
```
