```python
class pwskills2:
    
    def student_details(self,name,email_id,nmber):
        print(name,email_id,number)
        
    @staticmethod
    def mentor_mail_id(mail_id_mentor):
        print(mail_id_mentor)
        
    @staticmethod
    def mentor_class(list_mentor):
        pwskills2.mentor_mail_id(["gill30@gmail.com","kavan45@yahoo.com"])
        print(list_mentor)
        
    
    @classmethod
    def class_name(cls):
        cls.mentor_class(["gill","rohit"])
        
    def mentor(self,mentor_list):
        print(mentor_list)
        self.mentor_class(["krish","sudh"])
```

```python
pwskills2.class_name()
```

```python
pwskills2.mentor_class()
```

# PROPERTY DECORATORS GETTER SETTER AND DELETES 

```python
class pwskills6:
    
    def __init__(self,course_price,course_name):
        
        self.__course_price=course_price
        self.course_name=course_name
        
    @property
    def course_price_access(self):
        return self.__course_price
```

```python
pw=pwskills6(4500,"data science masters")
```

```python
pw._pwskills6__course_price
```

```python
pw.course_name
```

```python
pw.course_price_access
```
