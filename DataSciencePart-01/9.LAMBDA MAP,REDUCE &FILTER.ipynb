MAP REDUCE AND FILTER


```python
l3=[1,2,3,4,5]
```

```python
def test2(l):
    l1=[]
    for i in l:
        l1.append(i**2)
    return l1
    
```

```python
test2(l)
```

```python
def sq(x):
    return x**2
```

```python
list(map(sq,l3))
```

```python
list(map(lambda x:x**2,l3))
```

```python
l1=[1,2,3,4,5]
l2=[6,7,8,9,10]
list(map(lambda x,y:x+y,l1,l2))
```

```python
def add(x,y):
    return x+y
list(map(add,l1,l2))
```

```python
from functools import reduce
```

```python
l=[1,2,3,4,5]
```

```python
reduce(lambda x,y:x+y,l)
```

```python
reduce(lambda x,y,z:x+y+z,l)
```

```python
reduce(lambda x,y:x+y,[])
```

```python
reduce(lambda x,y:x+y,[l])
```

```python
reduce(lambda x,y:x if x>y else y,l)
```

```python
list(filter(lambda x:x%2==0,l))
```
