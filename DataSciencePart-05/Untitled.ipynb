## F test With Python

```python
## Evidence to say that the two population variance are not equal.
##Both the Performance is Different


worker1 = [18,19,22,25,27,28,41,45,51,55]
worker2 = [14,15,17,18,22,25,25,27,34] 
```

```python
##calcualte Ftest
import numpy as np
f_test = np.var(worker1)/np.var(worker2)
```

```python
f_test
```

```python
## Degree of Freedom
df1 = len(worker1) - 1
df2 = len(worker2) - 1
signfance_value = 0.05
```

```python
import scipy.stats as stat
```

```python
criticalvalue = stat.f.ppf(q = 1-signfance_value,dfn = df1,dfd = df2)
```

```python
criticalvalue
```

```python
if f_test>criticalvalue:
    print("Reject the NULL Hypothesis")
else:
    print("We fail to Reject the Null Hypothesis  ")
```
