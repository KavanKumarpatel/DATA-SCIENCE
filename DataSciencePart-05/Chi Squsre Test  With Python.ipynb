## Chi Square Test With Python

```python
import scipy.stats as stat
import numpy as np
```

```python
## No of hours students study in basis daily 
expected_data  = [8,6,7,9,6,9,7]
observed_data = [7,8,6,9,9,6,7]
```

```python
sum(expected_data),sum(observed_data)
```

```python
## Chi -square Goodness of fit Test
chisquare_test_statistic,p_value = stat.chisquare(observed_data,expected_data)
```

```python
chisquare_test_statistic,p_value
```

```python
##find the critical value
significane_value = 0.5
dof = len(expected_data)
print(dof)
critical_value = stat.chi2.ppf(1-significane_value)
```
