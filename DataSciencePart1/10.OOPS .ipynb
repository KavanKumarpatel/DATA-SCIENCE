```python
class bank_account:
    def __init__(self,balance):
        self.__balance=balance
        
    def deposit(self,amount):
        self.__balance=self.__balance+amount
    def withdraw(self,amount):
        if self.__balance>=amount:
            self.__balance=self.__balance-amount
            return True
        else :
            return False
    def get_balance(self):
        return self.__balance
```

```python
kavan=bank_account(1000)
```

```python
kavan.deposit(5000)
```

```python
kavan.get_balance()
```

```python
kavan.deposit(3400)
```

```python
kavan.get_balance()
```

```python
kavan.withdraw(10000)
```

```python
kavan.withdraw(34)
```

```python
kavan.get_balance()
```
