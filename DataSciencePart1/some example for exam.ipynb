```python
# polymorphism for class 

class collage1:
    def __init__(self,student_num,fee):
        self.student_num=student_num
        self.fee=fee
        
    def comapre(self):
        print("hard")
        
        
class collage2:
    
    def __init__(self,student_num,fee):
        self.student_num=student_num
        self.fee=fee
        
    
    def comapre(self):
        print("easy")
        
        
        
class collage3:
    
    def __init__(self,student_num,fee):
        
        self.student_num=student_num
        self.fee=fee
        
        
    def comapre(self):
        print("medium")
        
        
        
        
p1=collage1(78,80000)
p2=collage2(72,50000)
p3=collage3(74,70000)



for i in (p1,p2,p3):
    i.comapre()
```

```python
#custom exception class 

class SalaryNotInRange(Exception):
    
    def __init__(self,salary,message="salary is not in (5000,15000) range"):
        self.salary=salary
        self.message=message
        super().__init__(self.message)
        
        
salary = int(input("ENter your salary"))
if not 5000<salary <15000:
    raise SalaryNotInRange(salary)
```

```python
# factorial program 

def fact(n):
    if n==1:
        return 1
    
    else:
        f=n*fact(n-1)
    return f

num=int(input("Enter the value which you want factorial"))

result=fact(num)
print("The factorial of ",num," is :",result)
```
