```python
 from sklearn.preprocessing import LabelEncoder
```

```python
# create a simple dataframe with a categorical variable
df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
# create ab instance of labelencoder
encoder= LabelEncoder()
```

```python
encoder.fit_transform(df['color']
                     )
```

```python
# ordinal encoding
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

#create a simple datafarme with an ordinal variable
df = pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
## create an instance of the OrdinalEncoder class and fit_transform
encoder = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```
